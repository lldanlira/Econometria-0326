```python
import pandas as pd

# Datos de demanda (X) y precio (Y) en 10 ciudades
datos = {
    'Articulos_Demandados': [38, 62, 73, 85, 100, 122, 50, 70, 98, 123],
    'Precio': [50, 40, 35, 30, 20, 15, 50, 36, 25, 20]
}

# Crear el DataFrame
df = pd.DataFrame(datos)
display(df)
```

```python
#Creando un modelo de regresión lineal Años de Demanda vs Precios
x = sm.add_constant(df['Articulos_Demandados']) #Se utiliza para añadir una columna a la matriz de variables predictoras (x)
reg = sm.OLS(df['Precio'], x)
modelo = reg.fit()
print(modelo.summary())
```

```python
#Ahora vamos a crear la gráfica del modelo de regresión lineal
plt.figure(figsize=(10,6))
plt.scatter(df['Articulos_Demandados'], df['Precio'], label='Datos')
plt.plot(df['Articulos_Demandados'], modelo.predict(x), color='red', label='Regresión')
plt.title('Demanda de Artículos vs Precio (USD)')
plt.xlabel('Cantidad de Artículos Demandados')
plt.ylabel('Precio (USD)')
plt.grid(True)
plt.legend()
plt.show()
```
