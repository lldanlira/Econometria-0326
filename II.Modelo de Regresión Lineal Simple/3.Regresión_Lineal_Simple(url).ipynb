```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns #Sirve para crear gráficos atractivos en Python
import statsmodels.api as sm
```

```python
# Cargar URL de archivo
url = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv'
```

```python
#Cargar el url a un archivo en DataFrame
df = pd.read_csv(url)
```

```python
df.head()
```

```python
#Definamos datos
x=df['total_bill']
y=df['tip']
```

```python
# Crear el gráfico de dispersión
plt.figure(figsize=(10, 6)) # Opcional: ajusta el tamaño del gráfico para una mejor visualización
plt.scatter(df['total_bill'], df['tip'])
plt.title('Propina vs. Monto de la Cuenta')
plt.xlabel('Monto de la Cuenta ($)')
plt.ylabel('Propina ($)')
plt.grid(True)
plt.show()
```

```python
# Matriz de correlación
correlacion = df.corr(numeric_only=True)  # necesario para evitar errores con columnas no numéricas

# Visualización con mapa de calor
plt.figure(figsize=(8,5))
sns.heatmap(correlacion, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Matriz de Correlación para variables de clientes')
plt.show()
```
