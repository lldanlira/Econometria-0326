```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm
```

```python
#Cargar archivo de excel tipo (xlsx)
ruta = '/content/Salary_Data.xlsx'
```

```python
#Crear o cargar el archivo de excel en un DataFrame de panda (xlsx)
df_excel = pd.read_excel(ruta)
```

```python
#Cargar archivo de excel tipo (csv)
ruta2 = '/content/Salary_Data - Salary_Data.csv'
```

```python
#Crear o cargar el archivo de excel en un DataFrame de panda (csv)
df_csv = pd.read_csv(ruta2)
```

```python
#Creando un modelo de regresión lineal Años de experiencia vs Salario
x = sm.add_constant(df_excel['YearsExperience']) #Se utiliza para añadir una columna a la matriz de variables predictoras (x)
reg = sm.OLS(df_excel['Salary'], x)
modelo = reg.fit()
print(modelo.summary())
```

```python
#Ahora vamos a crear la gráfica del modelo de regresión lineal
plt.figure(figsize=(10,6))
plt.scatter(df_excel['YearsExperience'], df_excel['Salary'], label='Datos')
plt.plot(df_excel['YearsExperience'], modelo.predict(x), color='red', label='Regresión')
plt.title('Relación entre Años de Experiencia y Salario')
plt.xlabel('Años de Experiencia')
plt.ylabel('Salario (USD)')
plt.grid(True)
plt.legend()
plt.show()
```
