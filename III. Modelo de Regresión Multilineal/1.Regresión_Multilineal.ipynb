```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
```

```python
datos = {
    'Peso':[64, 71, 53, 67, 55, 58, 77, 57, 56, 51, 76, 68],
    'Estatura':[57, 59, 49, 62, 51, 50, 55, 48, 52, 42, 61, 57],
    'Edad':[8, 10, 6, 11, 8, 7, 10, 9, 10, 6, 12, 9]
}

df = pd.DataFrame(datos)
print(df)
```

```python
#Definiendo variables
y = df['Peso']
x = df[['Estatura', 'Edad']]
```

```python
#Creación del modelo de regresión lineal
x = sm.add_constant(x)
modelo = sm.OLS(y,x)
reg = modelo.fit()
resultado = reg.summary()
print(resultado)
```

```python
R2 = 0.709
R = np.sqrt(R2)
print(R)
```

Identificar:

* La ecuación del modelo de regresión multilineal.

$y = 0.8546x_1 +  1.5063x_2 +  3.6512 $

$Peso = 0.8546Edad +  1.5063Altura +  3.6512 $

* Coeficiente de Determinación: $R^2=0.709$

El 70.9% de la variabilidad del peso depende tanto de la edad como de la altura.

* Coeficiente de Correlación Lineal: $R = 0.8420$.
