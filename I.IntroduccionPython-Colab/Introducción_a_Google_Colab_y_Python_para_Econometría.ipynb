# **Librerías básicas para econometría**

🔹 NumPy (vectores y operaciones matemáticas)

```python
import numpy as np

# Crear un arreglo
x = np.array([1, 2, 3, 4, 5])
print("Arreglo:", x)

# Promedio y desviación estándar
print("Media:", np.mean(x))
print("Desviación estándar:", np.std(x))
print("Varianza:", np.var(x))
```

🔹 Pandas (manejo de datos tipo Excel)

```python
import pandas as pd

# Crear un DataFrame (tabla de datos)
datos = {
    "Año": [2020, 2021, 2022, 2023],
    "PIB": [1000, 1050, 1100, 1150],
    "Inflación": [3.5, 4.0, 5.2, 4.8]
}

pedro = pd.DataFrame(datos)
print(pedro)
```

🔹 Matplotlib (visualización de datos)

```python
import matplotlib.pyplot as plt

# Gráfico simple del PIB
plt.plot(pedro["Año"], pedro["PIB"], marker="o", linestyle = "--", color = "blue")
plt.title("PIB en el tiempo")
plt.xlabel("Año")
plt.ylabel("PIB (miles de millones)")
plt.show()
```

```python
print("Correlación:",np.corrcoef(pedro['Inflación'],pedro['PIB']))
```

Actividad práctica en clase

👉 Los estudiantes crearán un DataFrame con datos ficticios de salarios y consumo y harán:

Calcular la media, varianza y correlación.

Graficar salario vs consumo.
"Salario": [8, 10, 12, 15, 18]
 "Consumo": [7, 9, 11, 14, 16]

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
Valores = {
    "Salario": [8, 10, 12, 15, 18],
    "Consumo": [7, 9, 11, 14, 16]
}
pedro = pd.DataFrame(Valores)
```

```python
pedro
```

```python
#Media de las variables
print("Media de Consumo:", np.mean(pedro['Consumo']))
print("Media de Salario:", np.mean(pedro['Salario']))
```

```python
#Variación de las variables
print("Varianza de Consumo", np.var(pedro['Consumo']))
print("Varianza de Salario", np.var(pedro['Salario']))
```

```python
#Correlación entre variables
print("Correlación Consumo vs Salario", np.corrcoef(pedro['Consumo'], pedro['Salario']))
```
